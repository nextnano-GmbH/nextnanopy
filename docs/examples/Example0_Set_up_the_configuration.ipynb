# Example 0: Set up the configuration


## About this example: nextnano configuration

If you have a nextnano software license, the first thing you want to do is to set up the configuration so you can execute input files easily.

On the other hand, most of the nextnanopy features can be used without the need for a valid nextnano license, unless you want to execute nextnano input files via Python. We recommend to set up your configuration at the beginning.


## Quick setup (recommended)

If you only want to set up the nextnanopy configuration and are not interested in the details or advanced usage, you do not need to read the rest of this example. Just use our template.

__Our template (templates/config_nextnano.py in this GitHub repository) helps you to set up the configuration.__

1. Save the file locally. If you have installed nextnanopy by downloading and building the source from GitHub, the template already exists in your local copy of the repository. We recommend to make a copy of this configuration template and save it to another location to avoid your settings being overwritten by 'git pull'.
2. Adjust the paths to your license folder, output folder and installation folder.
3. Run the script, then the file .nextnanopy-config gets updated.

### Where do I find each path?

If you have activated your license via nextnanomat, the easiest way is to go to the nextnanomat/Tools/Options. Under the Simulation tab, you find the executable paths for each nextnano software (nextnano++, nextnano3, nextnano.NEGF).

Similarly, the database and license paths are in the Database and the License tabs.

---

Everything below explains what the configuration is and how it can be used: accessing it, setting each option manually, and executing input files with it. You only need it if you want the details or more advanced usage.

## What is the configuration?

The configuration includes the paths to the executables, the license files, the default output folder, etc. 

This configuration is stored as a file .nextnanopy-config and will be located at your home directory (usually C:\Users\Your_User\\.nextnanopy-config on Windows).

When you import nextnanopy for the first time, it will automatically generate the configuration file with a few default parameters. If a configuration file already exists, it won't modify it.

Once you have set up this configuration file, you do not need to worry about it anymore, except when you renew your license or update the nextnano software version.


## Let's import nextnanopy before setting up the configuration
In your Python console (e.g. in Spyder), type in:

````python
import nextnanopy as nn
````

This will generate a configuration file in your home directory.

## How can I access the configuration?

The configuration can be accessed via:

````python
nn.config
````

To get an option from a section (e.g. nextnano++, nextnano3, nextnano.NEGF or nextnano.MSB), you can use the get method:

````python
nn.config.get('nextnano++', 'exe')
````

To get all the options from a section, use:

````python
nn.config.get_options('nextnano++')
````

You can also get the location of the configuration file:

````python
nn.config.fullpath
````

## Side note: How can I set each option?

_Our config template does the following automatically. You do not have to do it by yourself!_

The command

```python
nextnanopy.config.set(section, option, value)
```
sets the option.
Please use absolute paths to avoid any possible problem and be aware of backslash or forward slash depending on your OS. We highly recommend to use 'r' in front of the path as shown below to avoid any syntax conflict in Python.

Note that after setting the options, you have to save the configuration by running

```python
nextnanopy.config.save()
```

You can also save the configuration elsewhere with a new name:

```python
nextnanopy.config.save(r'C:\new_path\random_name.nextnanopy-config')
```

In the last case, nextnanopy will always look for the configuration file in your HOME directory and NOT this one. Saving your own configuration file can be useful in some cases that we will show later in this Example.

````python
# This example configures nextnano++ only.
# To also set up nextnano3, nextnano.NEGF or nextnano.MSB, repeat the same
# nn.config.set(...) calls with the corresponding section name.
nn.config.set('nextnano++','exe',r'C:\Program Files\nextnano\2026_07_03\nextnano++\bin\nextnano++_Intel_64bit.exe')
nn.config.set('nextnano++','license',r'C:\Users\Heorhii\Documents\nextnano\License\License_nnp.lic')
nn.config.set('nextnano++','database',r'C:\Program Files\nextnano\2026_07_03\nextnano++\database\database.nnp')
nn.config.set('nextnano++','outputdirectory',r'C:\Users\Heorhii\Documents\nextnano\Output')

nn.config.save() #save permanently
````

````python
nn.config
````

## Execute input file with the default configuration

By loading an input file with the InputFile class (for more information, see Example1), it will automatically load the configuration file as well. You can access it via:

````python
my_input = nn.InputFile(r'..\..\templates\input files\DoubleQuantumWell_6nm_demo.nnp')
my_input.config
````

The execute() method runs the input file.


It will automatically detect the nextnano product (nextnano++, nextnano3, nextnano.NEGF or nextnano.MSB) and will load the corresponding configuration parameters.

````python
my_input.execute(show_log=False)
#  print the location of the output folder where the results are stored
print(my_input.folder_output)
````

## Execute with different parameters

### Method 1: use another configuration file

If you want to execute a specific input file with user-defined configuration file, you can do

```python
my_input = nextnanopy.InputFile('example.nnp', configpath=r'C:\new_path\random_name.nextnanopy-config')
my_input.config
```

This feature can be useful if you have different versions of nextnano installed and you want to use a given version for a set of input files.

### Method 2: without any configuration file

You can pass one or more specific arguments such as outputdirectory or threads:

````python
my_input.execute(outputdirectory=r'C:\Users\Heorhii\Downloads', threads=4, show_log=False)
# print the location of the output folder where the results are stored
print(my_input.folder_output)
````

### Method 3: set the configuration before creating the input file

You can also change the configuration in memory with `nn.config.set(...)` and then create the `InputFile` object. The input file copies the current configuration when it is created, so the values you set are the ones used by `execute()`.

Note the order: call `nn.config.set(...)` **before** `nn.InputFile(...)`. The configuration is bound at construction, so a change made afterwards would not reach an input file that already exists.

This only affects the current Python session. It is not saved to disk (there is no `nn.config.save()` call), so your `.nextnanopy-config` file stays unchanged.

````python
nn.config.set('nextnano++', 'outputdirectory', r'C:\Users\Heorhii\Downloads') # in memory only, not saved to disk

my_input = nn.InputFile(r'..\..\templates\input files\DoubleQuantumWell_6nm_demo.nnp')
my_input.execute(show_log=False)
# the new output directory is used, but .nextnanopy-config on disk stays unchanged
print(my_input.folder_output)
````

## Reset the configuration to default

You can reset the configuration to the default one.

Note that you need to save the file after the modification.

````python
nn.config.to_default() # not saved automatically
nn.config.save() # save the default values
print(nn.config)
````
